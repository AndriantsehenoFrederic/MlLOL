```python
df["datetime_utc"] = pd.to_datetime(df["datetime_utc"], utc=True, errors="coerce")

df = df.loc[
    df["datetime_utc"] >= pd.Timestamp("2026-01-01", tz="UTC") # à modifier en fonction de jusqu'à quand on veut des matchs
].copy()

print(f"{len(df)} matchs conservés")
print("Premier match :", df["datetime_utc"].min())
```
